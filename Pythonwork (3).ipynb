```python
#bank loan approval
country=input("Enter country:")
employment=input("Enter employment status:(yes/no)")
salary=float(input("Enter salary:"))
credit_score=int(input("Enter credit score:"))
if country=="India":
    if employment=="yes":
        if salary>=25000:
            if credit_score>=700:
                print("Loan Approved")
            else:
                print("Loan Rejected:Low credit score")
        else:
            print("Loan Rejected:Low salary")
    else:
        print("Loan Rejected:Not employed")
else:
    print("Loan Rejected:Country not eligible")          
```

```python
#college admission
state=input("Enter your state:")
marks=int(input("Enter your marks:"))
age=int(input("Enter your age:"))
qualification=input("Enter your qualification:")
if state=="Karnataka":
    if marks>=50:
        if age>=17:
            if qualification=="PUC":
                print("You are eligible for admission")
            else:
                print("Not eligible:Qualification required")
        else:
            print("Not eligible:Age is below 17")
    else:
        print("Not eligible:Marks are low")
else:
    print("Not eligible:State not eligible")     
```

```python
#online shopping
login=input("Are you logged in?(yes/no)")
product=input("is product available?(yes/no)")
payment=input("is payment successful?(yes/no)")
if login=="yes":
    if product=="yes":
        if payment=="yes":
            print("Order confirm")
        else:
            print("Order failed:Payment unsuccessful")
    else:
        print("Order failed:Product not available")
else:
    print("Please login first")   
```

```python
#Atm withdrawal
card=input("Is card valid?(yes/no):")
pin=input("Enter PIN:")
balance=float(input("Enter a/c balance:"))
amt=float(input("Enter withdrawal amt:"))
if card=="yes":
    if pin=="1234":
        if balance>=amt:
            print("Withdrawal successful")
        else:
            print("Remaining balance:",balance-amt)
    else:
        print("Incorrect PIN")
else:
    print("Invalid card")
        
```

```python
#insurance eligibility
country=input("Enter country:")
age=int(input("Enter your age:"))
income=float(input("Enter annual income:"))
medical=input("Do you meet medical criteria?(yes/no):")
if country=="India":
    if age>=18:
        if income>=200000:
            if medical=="yes":
                print("Eligible for insurance")
            else:
                print("Medical criteria not satisfied")
        else:
            print("Income is low")
    else:
        print("Age must be 18 or above")
else:
    print("Country not supported")

```

```python
#Employee promotion
dept=input("Enter department:")
experiance=int(input("Enter years of experiance:"))
performance=float(input("Enter performance percentage:"))
salary=float(input("Enter salary:"))
if dept=="IT":
    if experiance>=3:
        if performance>=75:
            if salary<=50000:
                print("Employee is eligible for promotion")
            else:
                print("Salary criteria is not satisfied")
        else:
            print("Low performance")
    else:
        print("Less experience")
else:
    print("Department criteria not satisfied")
        
```
