```python
# Question 1: Rectangle Area
# Write a function `rectangle_area(length, width)` that returns the area.
# rectangle_area(5, 3) -> 15
def rectangle_area(length,width):
    area=length*width
    return area
    
rectangle_area(5,3)
    
```

```python
# Question 2: Is Even
# Write a function `is_even(n)` that returns True if n is even.
# is_even(4) -> True
def is_even(n):
    return n % 2 == 0

is_even(4)    
```

```python
# Question 3: Sum of List
# Write a function `sum_list(numbers)` that returns the sum of a list.
# sum_list([1, 2, 3]) -> 6
def sum_list(numbers):
    return sum(numbers)
    
sum_list([2,4,6])    
```

```python
# Question 4: Reverse String
# Write a function `reverse_string(s)` that returns the reversed string.
# reverse_string("hello") -> "olleh"
def reverse_string(s):
    return s[::-1]

reverse_string("welcome")    
```

```python
# Question 5: Find Maximum
# Write a function `find_max(numbers)` that returns the largest number.
# find_max([3, 7, 2, 9, 4]) -> 9
def find_max(numbers):
    return max(numbers)

find_max([8,4,1,9,5])    
```

```python
# Question 6: Square a Number
# Write a function `square(n)` that returns n squared.
# square(5) -> 25
def square(n):
    return n ** 2

square(4)
    
```

```python
# Question 7: Is Palindrome
# Write a function `is_palindrome(s)` that returns True if the string
# reads the same forwards and backwards.
# is_palindrome("racecar") -> True
def is_palindrome(s):
    return s == s[::-1]

is_palindrome("level")    
```

```python
# Question 8: Count Vowels
# Write a function `count_vowels(s)` that returns the number of vowels
# (a, e, i, o, u) in a string.
# count_vowels("hello world") -> 3
def count_vowels(s):
    return sum(char.lower() in "aeiou" for char in s)

count_vowels("python")    
```

```python
# Question 9: Factorial
# Write a function `factorial(n)` that returns n! (n factorial).
# factorial(5) -> 120
def factorial(n):
    result = 1
    for number in range(2, n + 1):
        result *= number
    return result

factorial(5)
```

```python
# Question 10: Is Prime
# Write a function `is_prime(n)` that returns True if n is a prime number.
# is_prime(7) -> True
def is_prime(n):
    if n < 2:
        return False
    for divisor in range(2, int(n ** 0.5) + 1):
        if n % divisor == 0:
            return False
    return True

is_prime(5)    
```

```python
# Question 11: Celsius to Fahrenheit
# Write a function `celsius_to_fahrenheit(c)` that converts Celsius to
# Fahrenheit using the formula: F = C * 9/5 + 32
# celsius_to_fahrenheit(0) -> 32.0
def celsius_to_fahrenheit(c):
    return c * 9 / 5 + 32

celsius_to_fahrenheit(0)    
```

```python
# Question 12: Count Occurrences
# Write a function `count_occurrences(lst, item)` that returns how many
# times `item` appears in `lst`.
# count_occurrences([1, 2, 2, 3, 2], 2) -> 3
def count_occurrences(lst, item):
    return lst.count(item)

count_occurrences([2,7,3,2,5],2)    
```

```python
# Question 13: Average of List
# Write a function `average(numbers)` that returns the average of a list.
# average([2, 4, 6]) -> 4.0
def average(numbers):
    return sum(numbers) / len(numbers)

average([2,4,6])
```

```python
# Question 14: Remove Duplicates
# Write a function `remove_duplicates(lst)` that returns a list with
# duplicates removed, keeping the original order.
# remove_duplicates([1, 2, 2, 3, 1]) -> [1, 2, 3]
def remove_duplicates(lst):
    result = []
    for item in lst:
        if item not in result:
            result.append(item)
    return result

remove_duplicates([1, 2, 2, 3, 1])

    
```

```python
# Question 15: FizzBuzz
# Write a function `fizzbuzz(n)` that returns "Fizz" if n is divisible
# by 3, "Buzz" if divisible by 5, "FizzBuzz" if divisible by both,
# otherwise returns str(n).
# fizzbuzz(15) -> "FizzBuzz"
# fizzbuzz(9)  -> "Fizz"
# fizzbuzz(7)  -> "7"
def fizzbuzz(n):
    if n % 15 == 0:
        return "FizzBuzz"
    if n % 3 == 0:
        return "Fizz"
    if n % 5 == 0:
        return "Buzz"
    return str(n)

fizzbuzz(15)
fizzbuzz(9)
fizzbuzz(7)
```
